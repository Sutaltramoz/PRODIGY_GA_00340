```python
import random

# --- STEP 1: The Brain (Your training data) ---
# You can add much more text here to make it smarter!
text_data = """
Prodigy InfoTech provides internship opportunities in various domains. 
Generative AI is a field of Artificial Intelligence. 
AI can generate text, images, and music. 
Learning Python is the first step for an intern. 
Interns at Prodigy work on exciting tasks like GPT-2 and Stable Diffusion.
Python is great for AI development.
"""

# --- STEP 2: The Logic (Building the Chain) ---
words = text_data.split()
chain = {}

for i in range(len(words) - 1):
    current_word = words[i]
    next_word = words[i + 1]
    if current_word not in chain:
        chain[current_word] = []
    chain[current_word].append(next_word)

# --- STEP 3: Generate New Text ---
def generate_sentence(length=8):
    word = random.choice(list(chain.keys()))
    sentence = [word]
    for _ in range(length - 1):
        if word in chain:
            word = random.choice(chain[word])
            sentence.append(word)
        else:
            break
    return " ".join(sentence)

print("--- PRODIGY TASK 03: MARKOV CHAIN OUTPUT ---")
for i in range(3):
    print(f"Generated {i+1}: {generate_sentence()}")
```
